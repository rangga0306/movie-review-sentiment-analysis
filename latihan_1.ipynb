```python
def hitung_rata_rata(nilai_list):
  total = sum(nilai_list)
  jumlah = len(nilai_list)
  return total/jumlah

nilai_ujian = [85.00,77.75,82.75,80.80,79.63,81.67,73.60,78.00]
print(f"rata-rata: {hitung_rata_rata(nilai_ujian)}")
```

```python
def cek_kelulusan(nilai):
  if nilai >=80:
    return "A"
  elif nilai >=70:
    return "B"
  elif nilai >=60:
    return "C"
  elif nilai >=50:
    return "D"
  else:
    return "E"

nilai_mahasiswa = [85.00,77.75,82.75,80.80,79.63,81.67,73.60,78.00]
for nilai in nilai_mahasiswa:
  print(f"nilai{nilai}:{cek_kelulusan(nilai)}")
```
